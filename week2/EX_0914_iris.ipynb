```python
# Decision Tree
```

```python
import pandas as pd

from sklearn.model_selection import train_test_split

from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import classification_report
```

```python
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/iris.csv")
```

```python
X = df.drop("species", axis=1)

y = df["species"] #label(정답)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
model = DecisionTreeClassifier()
model.fit(X_train, y_train)
#모델을 만들고 정답과 데이터를 넘겨줘서 학습시킴
```

```python
pred_dt = model.predict(X_test) #예측값
```

```python
print(classification_report(y_test, pred_dt)) #정답과 예측값 비교
```

```python
# Random Forest
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model = RandomForestClassifier()

model.fit(X_train, y_train)
```

```python
pred_rf = model.predict(X_test)
```

```python
print(classification_report(y_test, pred_rf))
```

```python
# Logistic Regression
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression(max_iter=200)

model.fit(X_train, y_train)
```

```python
pred_lr = model.predict(X_test)
```

```python
print(classification_report(y_test, pred_lr))
```

```python
# KNN
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
model = KNeighborsClassifier()

model.fit(X_train, y_train)
```

```python
pred_kn = model.predict(X_test)
```

```python
print(classification_report(y_test, pred_kn))
```
