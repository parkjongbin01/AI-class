## 2. 데이터 불러오기

```python
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/wine.csv")

df
```

## 3. 데이터 구조 확인

```python
print("데이터 크기:", df.shape)

print("\n처음 5개 데이터")
display(df.head())

print("\n변수 이름")
print(df.columns.tolist())

print("\n자료형")
print(df.dtypes)

print("\n데이터 정보")
df.info()
```

## 5. 분석 변수 지정

```python
numeric_cols = [
    'Alcohol',
    'Malic.acid',
    'Ash',
    'Acl',
    'Mg',
    'Phenols',
    'Flavanoids',
    'Nonflavanoid.phenols',
    'Proanth',
    'Color.int',
    'Hue',
    'OD',
    'Proline'
]

target_col = 'Wine'
```

## 6. 레이블 종류와 개수 확인

```python
print("레이블 종류")
print(df[target_col].unique())

print("\n레이블별 개수")
print(df[target_col].value_counts())
```

```python
# 레이블 갯수 확인 (숫자)

df['Wine'].value_counts()
```

### 레이블 개수 시각화

```python
plt.figure(figsize=(6, 4))

ax = sns.countplot(
    data=df,
    x=target_col
)

for container in ax.containers:
    ax.bar_label(container)

plt.title("Number of Samples by Wine")
plt.xlabel("Wine")
plt.ylabel("Count")

plt.tight_layout()
plt.show()
```

## 7. 기술통계량 확인

```python
display(df[numeric_cols].describe().round(3))
```

## 8. 품종별 평균과 표준편차 확인

```python
print("품종별 평균")
display(
    df.groupby(target_col)[numeric_cols]
      .mean()
      .round(3)
)

print("품종별 표준편차")
display(
    df.groupby(target_col)[numeric_cols]
      .std()
      .round(3)
)
```

## 9. 각 변수의 전체 분포 확인

```python
for col in numeric_cols:

    plt.figure(figsize=(7, 4))

    sns.histplot(
        data=df,
        x=col,
        bins=20,
        kde=True
    )

    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")

    plt.tight_layout()
    plt.show()
```

## 10. 품종별 변수 분포 비교

```python
for col in numeric_cols:

    plt.figure(figsize=(7, 4))

    sns.histplot(
        data=df,
        x=col,
        hue=target_col,
        kde=True,
        element="step",
        stat="density",
        common_norm=False
    )

    plt.title(f"{col} Distribution by Species")

    plt.tight_layout()
    plt.show()
```

## 11. Box Plot

```python
for col in numeric_cols:

    plt.figure(figsize=(7, 4))

    sns.boxplot(
        data=df,
        x=target_col,
        y=col
    )

    plt.title(f"{col} by Species")

    plt.tight_layout()
    plt.show()
```

## 12. Violin Plot

```python
for col in numeric_cols:

    plt.figure(figsize=(7, 4))

    sns.violinplot(
        data=df,
        x=target_col,
        y=col,
        inner="quartile"
    )

    plt.title(f"{col} Distribution by Species")

    plt.tight_layout()
    plt.show()
```

## 13. Pair Plot

```python
sns.pairplot(
    df,
    vars=numeric_cols,
    hue=target_col,
    diag_kind="hist"
)

plt.show()
```

## 15. Flavanoids와 Proline의 관계


Flavanoids와 Proline는 품종별 군집이 비교적 뚜렷하게 나타나는 변수입니다.

```python
plt.figure(figsize=(7, 5))

sns.scatterplot(
    data=df,
    x='Flavanoids',
    y='Proline',
    hue=target_col,
    s=70
)

plt.title("Flavanoids vs Proline")
plt.xlabel("Flavanoids")
plt.ylabel("Proline")

plt.tight_layout()
plt.show()
```

## 16. Ash와 Mg의 관계

Flavanoids vs Proline와 비교하기 위해 Ash와 Mg의 관계도 확인합니다.

Ash와 Mg에서는 품종별 데이터가 상대적으로 더 많이 겹치는 것을 볼 수 있습니다.

```python
plt.figure(figsize=(7, 5))

sns.scatterplot(
    data=df,
    x='Ash',
    y='Mg',
    hue=target_col,
    s=70
)

plt.title("Ash vs Mg")
plt.xlabel("Ash")
plt.ylabel("Mg")

plt.tight_layout()
plt.show()
```

## 17. IQR을 이용한 이상치 탐지

```python
for col in numeric_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower) |
        (df[col] > upper)
    ]

    print(
        f"{col:20s} : "
        f"{len(outliers)}개 "
        f"(Lower={lower:.2f}, Upper={upper:.2f})"
    )
```

### 실제 이상치 데이터 확인

```python
for col in numeric_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower) |
        (df[col] > upper)
    ]

    if len(outliers) > 0:
        print(f"\n[{col} 이상치]")
        display(outliers[[col, target_col]])
```

## 18. 품종별 평균값 비교

```python
mean_long = (
    df.groupby(target_col)[numeric_cols]
      .mean()
      .reset_index()
      .melt(
          id_vars=target_col,
          var_name="Feature",
          value_name="Mean"
      )
)

plt.figure(figsize=(14, 5))

sns.barplot(
    data=mean_long,
    x="Feature",
    y="Mean",
    hue=target_col
)

plt.title("Mean Feature Values by Wine")
plt.xlabel("Feature")
plt.ylabel("Mean (Log Scale)")

plt.xticks(rotation=20)

plt.yscale('log')
#로그스케일로 변경

plt.tight_layout()
plt.show()
```
